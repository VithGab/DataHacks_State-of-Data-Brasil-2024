```python
SALARY_MAP = {
    "Menos de R$ 1.000/mês": 750,
    "de R$ 1.001/mês a R$ 2.000/mês": 1500,
    "de R$ 2.001/mês a R$ 3.000/mês": 2500,
    "de R$ 3.001/mês a R$ 4.000/mês": 3500,
    "de R$ 4.001/mês a R$ 6.000/mês": 5000,
    "de R$ 6.001/mês a R$ 8.000/mês": 7000,
    "de R$ 8.001/mês a R$ 12.000/mês": 10000,
    "de R$ 12.001/mês a R$ 16.000/mês": 14000,
    "de R$ 16.001/mês a R$ 20.000/mês": 18000,
    "de R$ 20.001/mês a R$ 25.000/mês": 22500,
    "de R$ 25.001/mês a R$ 30.000/mês": 27500,
    "de R$ 30.001/mês a R$ 40.000/mês": 35000,
    "Acima de R$ 40.001/mês": 45000
}

df["salario_ponto_medio"] = df["2.h_faixa_salarial"].map(SALARY_MAP)

display(df[["2.h_faixa_salarial","salario_ponto_medio"]].head())

print("Média Salarial:", df["salario_ponto_medio"].mean())
print("Mediana Salarial:", df["salario_ponto_medio"].median())
```

### Tratamento das Faixas Salariais

A variável de salário foi coletada em faixas de renda (ex.: *"de R$ 4.001/mês a R$ 6.000/mês"*), impossibilitando cálculos estatísticos como:
* Média
* Mediana
* Correlação
* Regressão

Para permitir análises quantitativas, cada faixa foi convertida para seu **ponto médio**. Essa abordagem foi escolhida por representar uma estimativa equilibrada do salário dentro do intervalo, evitando a subestimação do limite inferior e a superestimação do limite superior.

#### Exemplos de Conversão:
* De R$ 4.001 a R$ 6.000 $\rightarrow$ **R$ 5.000**
* De R$ 8.001 a R$ 12.000 $\rightarrow$ **R$ 10.000**

> ⚠️ **Faixa aberta:** Para a opção *"Acima de R$ 40.001/mês"*, foi adotado o valor de **R$ 45.000** como uma aproximação conservadora.

### Tratamento das Ferramentas Utilizadas

As linguagens e ferramentas utilizadas pelos profissionais não aparecem em uma única coluna, mas sim em diversas colunas binárias (0 = não utiliza, 1 = utiliza).

Essa estrutura foi mantida por representar corretamente respostas múltiplas. Um profissional pode utilizar simultaneamente Python, SQL e Power BI, por exemplo.

Os valores ausentes foram convertidos para 0, assumindo que a ausência de marcação indica que a ferramenta não foi selecionada pelo participante.

Além disso, foi criada uma variável denominada "qtd_ferramentas", representando a quantidade total de ferramentas utilizadas por cada respondente.

Essa abordagem permite análises de frequência e comparações entre tecnologias utilizadas no mercado.

```python
cargo_map = {
    "Analista de Dados/Data Analyst":
        "Analista de Dados",

    "Analista de BI/BI Analyst":
        "Analista de BI",

    "Analista de Negócios/Business Analyst":
        "Analista de Negócios",

    "Cientista de Dados/Data Scientist":
        "Cientista de Dados",

    "Engenheiro de Dados/Data Engineer/Data Architect":
        "Engenheiro de Dados",

    "Analytics Engineer":
        "Analytics Engineer",

    "Engenheiro de Machine Learning/ML Engineer/AI Engineer":
        "ML Engineer",

    "Desenvolvedor/ Engenheiro de Software/ Analista de Sistemas":
        "Desenvolvedor",

    "Professor/Pesquisador":
        "Professor/Pesquisador",

    "Estatístico":
        "Estatístico"
}

df["cargo_padronizado"] = (
    df["cargo_original"]
    .map(cargo_map)
    .fillna("Outros")
)

display(
    df["cargo_padronizado"]
    .value_counts()
)
```

Optou-se por não utilizar algoritmos automáticos de similaridade textual, pois eles poderiam agrupar cargos distintos de forma incorreta.

# **Análise Univariada das Variáveis**

---



A variável salário é a principal do estudo, pois a pergunta central busca identificar os fatores que mais influenciam a remuneração dos profissionais de dados no Brasil.

Será analisada sua distribuição, tendência central e dispersão para compreender o comportamento salarial dos respondentes.

---



```python
def resumo_estatistico(df, coluna):

    print("="*60)
    print(f"ANÁLISE DA VARIÁVEL: {coluna}")
    print("="*60)

    serie = df[coluna].dropna()

    print("Média:", round(serie.mean(),2))
    print("Mediana:", round(serie.median(),2))
    print("Moda:", serie.mode().iloc[0])
    print("Variância:", round(serie.var(),2))
    print("Desvio padrão:", round(serie.std(),2))
    print("Amplitude:", round(serie.max()-serie.min(),2))
    print("Assimetria:", round(serie.skew(),2))
    print("Curtose:", round(serie.kurtosis(),2))
```

função criada para evitar a repetição do mesmo código

## Salário × Senioridade

---



A senioridade é tradicionalmente um dos principais fatores associados à remuneração.

O objetivo desta análise é verificar se profissionais mais seniores apresentam salários mais elevados.

```python
salario_senioridade = (
    df.groupby("senioridade")["salario_ponto_medio"]
      .mean()
      .sort_values()
)

display(salario_senioridade)
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    data=df,
    x="senioridade",
    y="salario_ponto_medio",
    estimator=np.mean,
    palette=purple_pink_palette
)

plt.title("Salário Médio por Senioridade")
plt.xlabel("Senioridade")
plt.ylabel("Salário Médio (R$)")

plt.xticks(rotation=45)
plt.show()
```

```python
senioridade_map = {
    "Júnior":1,
    "Pleno":2,
    "Sênior":3,
    "Especialista":4
}

df["senioridade_num"] = (
    df["senioridade"]
    .map(senioridade_map)
)

dados = df[
    ["senioridade_num","salario_ponto_medio"]
].dropna()

corr, p = spearmanr(
    dados["senioridade_num"],
    dados["salario_ponto_medio"]
)

print("Spearman:", round(corr,3))
print("P-value:", p)
```



A correlação de Spearman foi utilizada porque senioridade é uma variável ordinal.

Valores positivos indicam que salários tendem a aumentar conforme o profissional avança de nível.

Quanto mais próximo de 1 estiver o coeficiente, mais forte é essa relação.

## Salário × Experiência em Dados

O tempo de experiência é frequentemente apontado como um dos principais determinantes salariais.

Nesta etapa será analisado se profissionais mais experientes recebem salários maiores.

```python
exp_salario = (
    df.groupby("exp_dados")["salario_ponto_medio"]
      .mean()
      .sort_values()
)

display(exp_salario)
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(
    data=df,
    x="exp_dados",
    y="salario_ponto_medio"
)

plt.title("Salário por Tempo de Experiência")
plt.xlabel("Experiência")
plt.ylabel("Salário")

plt.xticks(rotation=45)
plt.show()
```

```python
exp_map = {
    "Menos de 1 ano":1,
    "1 a 2 anos":2,
    "3 a 4 anos":3,
    "5 a 6 anos":4,
    "7 a 10 anos":5,
    "Mais de 10 anos":6
}

df["exp_num"] = (
    df["exp_dados"]
    .map(exp_map)
)
```

```python
dados = df[
    ["exp_num","salario_ponto_medio"]
].dropna()

corr, p = spearmanr(
    dados["exp_num"],
    dados["salario_ponto_medio"]
)

print("Spearman:", round(corr,3))
print("P-value:", p)
```



A experiência é uma variável ordinal, portanto a correlação de Spearman é mais apropriada.

Uma correlação positiva indica tendência de crescimento salarial conforme aumenta o tempo de atuação na área.
